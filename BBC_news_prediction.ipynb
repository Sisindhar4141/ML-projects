```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score ,f1_score,classification_report
data = pd.read_csv('bbc_text_cls new.csv')
df = pd.DataFrame(data)
print(df.columns)
vectorizer = TfidfVectorizer(stop_words='english',max_features=50000,ngram_range=(1, 2),sublinear_tf=True)
x = vectorizer.fit_transform(df['text'])
y = df['labels']
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
model = SVC(kernel='rbf',C=10,gamma='scale',probability=True)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('F1 score:', f1_score(y_test, y_pred, average='weighted'))
results = pd.DataFrame({'Actual_values': y_test.values, 'Predicted_values': y_pred})
print(results.head())
```

```python
new_headlines = 'he is elected as the new mayor'
headline_vector = vectorizer.transform([new_headlines])
prediction = model.predict(headline_vector)
print(f'Headline: "{new_headlines}"')
print(f'Prediction: {prediction}\n')
print('Confidence scores:')
for headline, label in zip(new_headlines, prediction):
    print(f'Headline: {headline} ---> Type: {label}')

print(classification_report(y_test, y_pred))
print("F1_score is :", f1_score(y_test, y_pred, average='weighted'))
```
